**Distanzen**

Dieses Notebook illustriert, wie man eine eigene domänspezifisische Distanzfunktion implementieren und benutzen kann.

```python
import numpy as np
```

```python
x1 = [2, 5, 10]
x2 = [8, 3, 12]
```

```python
x1 = np.array(x1)
x2 = np.array(x2)
```

```python
def manhattan_distance(x1, x2):
  return np.sum(np.abs(x1-x2))
```

```python
historische_daten = np.array( [ [42,38,0],
                                [25,60,1],
                                [32,90,2],
                                [45,80,0] ])
```

```python
historische_daten[1]
```

```python
neuer_kunde = np.array( [ 27, 95, 2] )
```

```python
# Eine "eigene" domänspezifische Distanzfunktion
def meine_distanz(x1, x2):
  wichtigkeit = [1, 3]
  d = 0
  for i in range(2):
    d = d + wichtigkeit[i]*abs(x1[i]-x2[i])
  if x1[2] != x2[2]:
    d = d + 5
  return d
```

```python
# Wir bestimmen, welcher Kunde in den historischen Daten der ähnlichste zu dem
# neuen Kunden ist.
distanzen = []
for i in range(len(historische_daten)):
  distanzen.append(meine_distanz(neuer_kunde,historische_daten[i]))
np.argmin(distanzen)
```
